# Sheet 5 - Maximum likelihood estimation in point-process data

A radio survey detects $N$ Fast Radio Bursts above a fluence (the time-integrated flux density of a transient, in this case the radio burst) threshold $S_{\min}$. Define the dimensionless fluence
$$
x \equiv S/S_{\min} \ge 1.
$$
Assume the bursts are a realisation of an inhomogeneous Poisson process on $x \in [1,\infty)$ with differential source count
$$
\phi(x;\boldsymbol{\theta}) = \frac{\mathrm{d}N}{\mathrm{d}x} = A\,x^{-\gamma}, \qquad \gamma > 1,
$$
where $\boldsymbol{\theta} = (A,\gamma)^\top$. The data are the observed fluences $\{x_i\}_{i=1}^N$ and the total number of detections $N$. The unbinned Poisson-process log-likelihood is
$$
\ell(\boldsymbol{\theta}) = -\bar N(\boldsymbol{\theta}) + \sum_{i=1}^N \ln\phi(x_i;\boldsymbol{\theta}),
\qquad
\bar N(\boldsymbol{\theta}) = \int_1^\infty \phi(x;\boldsymbol{\theta})\,\mathrm{d}x.
$$
You can calculate everything (apart from the visualisation) in this exercise **analytically**!


a. Evaluate $\bar N(\boldsymbol{\theta})$ and write $\ell(\boldsymbol{\theta})$ explicitly. Why does $\gamma > 1$ matter?

b. Derive the MLEs $\hat A$ and $\hat\gamma$ in closed form.

c. Is $\hat\gamma$ unbiased? Compute $\langle \hat\gamma\rangle$ exactly and provide a bias-corrected estimator.

d. Derive the Poisson-process Fisher identity
$$
F_{ab} = \int \phi\,\partial_a\ln\phi\,\partial_b\ln\phi\,\mathrm{d}x,
$$
and use it to compute $F(\boldsymbol{\theta})$. Comment on parameter dependence and the off-diagonal term. *Hint*: use the Poisson-process identity $\langle\sum_i g(x_i)\rangle = \int \phi(x)g(x)\,\mathrm{d}x$.

e. Invert $F$. Read off the variances of $\hat A$, $\hat\gamma$, and their correlation coefficient. Show that the latter is parameter-independent.

f. Reparametrise to $(\bar N,\gamma)$ and show that the Fisher matrix becomes diagonal. Interpret which coupling is physical and which is a parametrisation artefact.

g. Visualise the likelihood around the best fit. Compute $\Delta\chi^2 \equiv -2\,[\ell(A,\gamma)-\ell(\hat A,\hat\gamma)]$ on a grid in $(A,\gamma)$. Use these numbers:
   - $N=100$
   - $\sum_i \ln x_i = 66.7$
   - $\hat\gamma = 1 + N/(\sum_i\ln x_i) \approx 2.50$
   - $\hat A = N(\hat\gamma-1) \approx 150$

   i. Make two 1D plots (profile slices): $\Delta\chi^2(A,\hat\gamma)$ vs. $A$ and $\Delta\chi^2(\hat A,\gamma)$ vs. $\gamma$. Mark the 68% and 95% 1D thresholds ($\Delta\chi^2=1,4$).
   
   ii. Make a 2D contour plot in the $(A,\gamma)$ plane and draw the 68% and 95% confidence contours ($\Delta\chi^2=2.30,5.99$). Compare the contour tilt to your Fisher-matrix correlation from part (e).
