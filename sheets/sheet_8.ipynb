# 8.1 Optimal predictor and risk decomposition

The following exercises go through the learning problem from the notes in a specific setting. Consider the data-generating process
$$
x \sim \mathcal{N}(0, 1), \qquad y = x^2 + \varepsilon, \qquad \varepsilon \sim \mathcal{N}(0, \sigma_\varepsilon^2),
$$
where $\varepsilon$ is independent of $x$.

a. Compute the optimal predictor $h^\star(x) = \langle y \mid x \rangle$ (cf. Eq. 7.6 in the notes) and the irreducible noise $\sigma^2 \equiv \langle \mathrm{Var}(y \mid x) \rangle_{f(x)}$ (cf. Eq. 7.10).

b. Consider the affine hypothesis class $\mathcal{H} = \{h_{\boldsymbol{w}}(x) = w_0 + w_1 x \mid \boldsymbol{w} \in \mathbb{R}^2\}$. Using the moments $\langle x^k \rangle = 0$ for $k$ odd and $\langle x^2 \rangle = 1$, $\langle x^4 \rangle = 3$ for $x \sim \mathcal{N}(0,1)$, compute the risk
$$
R(h_{\boldsymbol{w}}) = \big\langle (y - h_{\boldsymbol{w}}(x))^2 \big\rangle_{f(x,y)}
$$
explicitly as a function of $\boldsymbol{w}$ and find the minimum $\boldsymbol{w}^\star = (w_0^\star, w_1^\star)^\top$. What predictor does $h_{\boldsymbol{w}^\star}$ correspond to?

c. Using the two-term risk decomposition (Eq. 7.9 in the notes), verify
$$
R(h_{\boldsymbol{w}^\star}) = \sigma^2 + \big\langle (h^\star(x) - h_{\boldsymbol{w}^\star}(x))^2 \big\rangle_{f(x)}.
$$
Compute each term explicitly and interpret the approximation error: what does it measure in terms of the hypothesis class $\mathcal{H}$?

# 8.2 Gradient of the logistic regression loss

For $N$ training examples $\{(\boldsymbol{x}_i, y_i)\}_{i=1}^N$ with $y_i \in \{0,1\}$, the empirical cross-entropy loss is
$$
\hat{\mathcal{L}}(\boldsymbol{w}) = \frac{1}{N}\sum_{i=1}^N \ell\big(y_i,\,h_{\boldsymbol{\theta}}(\boldsymbol{x}_i)\big),
$$
where $\ell$ is the cross-entropy loss (Eq. 7.16 in the notes) and $h_{\boldsymbol{\theta}}(\boldsymbol{x}) = \mathrm{sig}(\boldsymbol{w}^\top \boldsymbol{x})$ (Eq. 7.14).

a. Show that the sigmoid satisfies
$$
\frac{\mathrm{d}}{\mathrm{d}t}\,\mathrm{sig}(t) = \mathrm{sig}(t)\,\big(1 - \mathrm{sig}(t)\big).
$$

b. Using the chain rule and part a., compute $\nabla_{\boldsymbol{w}}\hat{\mathcal{L}}(\boldsymbol{w})$ and show that it takes the compact form
$$
\nabla_{\boldsymbol{w}}\hat{\mathcal{L}}(\boldsymbol{w}) = \frac{1}{N}\sum_{i=1}^N \big(h_{\boldsymbol{\theta}}(\boldsymbol{x}_i) - y_i\big)\,\boldsymbol{x}_i.
$$
*Hint:* let $u_i = \boldsymbol{w}^\top\boldsymbol{x}_i$ and compute $\partial\hat{\mathcal{L}}/\partial u_i$ first.

c. Write down the gradient descent update rule $\boldsymbol{w} \leftarrow \boldsymbol{w} - \eta\,\nabla_{\boldsymbol{w}}\hat{\mathcal{L}}$. For an example $i$ with $y_i = 0$ but $h_{\boldsymbol{\theta}}(\boldsymbol{x}_i) \approx 1$, describe the direction and magnitude of the resulting update and explain intuitively why it has the correct effect.

# 8.3 Overfitting and the bias–variance tradeoff

This exercise asks you to study overfitting and the bias–variance tradeoff through polynomial regression on a real dataset. The data are provided in three files in the data directory:

- `sheet_8_data.txt` — 30 labelled points $(x_i, y_i)$, your training set.
- `sheet_8_test.txt` — 1000 held-out points drawn from the same distribution.
- `sheet_8_resamples.txt` — 500 independently drawn training sets of size 30 (columns: `set_id`, `x`, `y`), for part c.

The true data-generating process is not disclosed. A degree-$k$ polynomial hypothesis class fits the model
$$
h_{\boldsymbol{w}}(x) = \sum_{j=0}^{k} w_j\, x^j
$$
by minimising the empirical squared-error risk (ordinary least squares).

a. Load `sheet_8_data.txt`. Fit polynomials of degree $k\in\{0,1,3,5,9\}$ and plot the fitted curves together with the training data. Describe qualitatively what you observe for low and high $k$. What do you think the true underlying function looks like?

b. Using the 30 training points and the 1000 test points from `sheet_8_test.txt`, fit polynomials of degree $k=0,1,\ldots,15$ and evaluate the mean squared error on both sets. Plot the training error and test error as functions of $k$ on a common axis. Identify the degree that minimises the test error and explain the observed U-shape.

c. Load `sheet_8_resamples.txt`. For each of the $M=500$ training sets and each degree $k\in\{1,5,9\}$, fit a degree-$k$ polynomial and record the prediction at $x_0=0.5$.

1. Estimate the variance of predictions at $x_0$ across the 500 fits:
$$
\widehat{\mathrm{var}} = \frac{1}{M}\sum_{m=1}^M \left(\hat{h}^{(m)}(x_0) - \frac{1}{M}\sum_{m'=1}^M\hat{h}^{(m')}(x_0)\right)^2.
$$
2. Using the test set, estimate the expected squared prediction error at $x_0$ by selecting all test points with $|x - x_0| < 0.05$ and computing $\frac{1}{M}\sum_m (\bar{y}_0 - \hat{h}^{(m)}(x_0))^2$, where $\bar{y}_0$ is the mean test label in that neighbourhood (a proxy for $h^\star(x_0)$).
3. Estimate the irreducible noise $\hat{\sigma}^2$ as the variance of test labels in the neighbourhood of $x_0$.
4. Verify numerically that
$$
\text{expected squared error} \approx \widehat{\mathrm{bias}}^2 + \widehat{\mathrm{var}} + \hat{\sigma}^2.
$$
Display the components in a table for each $k$ and comment on how they change with polynomial degree.

d. Repeat part b for training set sizes $N\in\{10,30,100\}$ by subsampling `sheet_8_resamples.txt`. How does increasing $N$ affect the optimal polynomial degree and the minimum achievable test error?
